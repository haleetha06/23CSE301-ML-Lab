```python
# CH.SC.U4CSE24157
# HALITHA BEGAM S
# K NEAREST NEIGHBORS
# Exercise 2

import pandas as pd
import numpy as np

# 1. Create the dataset

data = {
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class': ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

# 2. Define the new entry

new_brightness = 55
new_saturation = 60

print("\nNew Entry:")
print("Brightness:", new_brightness)
print("Saturation:", new_saturation)

# 3. Calculate Euclidean distance

df['Distance'] = np.sqrt(
    (df['Brightness'] - new_brightness) ** 2 +
    (df['Saturation'] - new_saturation) ** 2
)

# 4. Display the distances

print("\nEuclidean Distances:")
print(df[['Brightness', 'Saturation', 'Class', 'Distance']])

# 5. Find the nearest data point

nearest = df.loc[df['Distance'].idxmin()]

print("\nNearest Data Point:")
print(nearest)

# 6. Display the nearest class

print("\nNearest Class:", nearest['Class'])
```
