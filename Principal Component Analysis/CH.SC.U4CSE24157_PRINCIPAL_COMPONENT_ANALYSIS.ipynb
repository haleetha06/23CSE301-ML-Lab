```python
# CH.SC.U4CSE24157
# HALITHA BEGAM S
# PRINCIPAL COMPONENT ANALYSIS
# Exercise 2

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create the dataset

data = {
    'Variable': [
        'Climate', 'Housing', 'Health', 'Crime',
        'Transportation', 'Education', 'Arts',
        'Recreation', 'Economy'
    ],

    'PC1': [
        0.190, 0.544, 0.782, 0.365,
        0.585, 0.394, 0.985, 0.520, 0.142
    ],

    'PC2': [
        0.017, 0.020, -0.605, 0.294,
        0.085, -0.273, 0.126, 0.402, 0.150
    ],

    'PC3': [
        0.207, 0.204, 0.144, 0.585,
        0.234, 0.027, -0.111, 0.519, 0.239
    ]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

# 2. Find the most important variable for each principal component

for pc in ['PC1', 'PC2', 'PC3']:

    index = df[pc].abs().idxmax()

    variable = df.loc[index, 'Variable']
    loading = df.loc[index, pc]

    print(f"\nMost important variable for {pc}:")
    print("Variable:", variable)
    print("Loading:", loading)

# 3. Display all principal components

print("\nPrincipal Component Loadings:")
print(df.set_index('Variable'))

# 4. Visualize the loadings

plt.figure(figsize=(10, 6))

sns.heatmap(
    df.set_index('Variable'),
    annot=True,
    fmt=".3f",
    cmap="Blues",
    center=0
)

plt.title("Principal Component Loadings")
plt.xlabel("Principal Components")
plt.ylabel("Variables")
plt.show()
```
